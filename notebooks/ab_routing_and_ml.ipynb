# Routing experiment and p95 forecast

All-Cloud is the control. Hybrid can finish on VeriFine or fall back to cloud. Demo costs and labels model a randomized scenario. Failures stay in their assigned arm. The primary endpoint is hybrid minus control Macro-F1 and the non-inferiority margin is −1.5 percentage points. SRM uses the declared 50/50 split, and Welch and Mann–Whitney tests are secondary. The planning estimate assumes equal true F1 and 80% power. The latency forecast uses a chronological holdout.


```python
import os
import sys
from pathlib import Path

os.environ.setdefault("OMP_NUM_THREADS", "2")
sys.path.insert(0, str(Path.cwd() if (Path.cwd() / "analysis").exists() else Path.cwd().parent))
import matplotlib.pyplot as plt
import pandas as pd
from IPython.display import display

from analysis.routing import analyze_experiment, fit_latency, load_experiment

events = load_experiment("routing-demo-42", "synthetic")
ab = analyze_experiment(events)
display(pd.Series(ab, name="Synthetic routing experiment"))

model_report, predictions = fit_latency(events)
display(pd.Series(model_report, name="Chronological model evaluation"))
daily = predictions.groupby("event_date")[["p95_ms", "predicted_p95_ms"]].mean()
ax = daily.plot(figsize=(10, 4), color=["#246b7e", "#ce7050"])
ax.set(title="Holdout mean of workflow-level p95s", ylabel="Milliseconds", xlabel="")
ax.legend(["Observed group p95", "Forecast group p95"])
plt.tight_layout()
plt.show()
```
